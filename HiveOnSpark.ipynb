```python
import pyspark
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("HiveExampleOnSpark").enableHiveSupport().getOrCreate()
```

```python
spark.sql("""CREATE TABLE employee (id INT, name STRING, age INT, salary DOUBLE)""");
```

```python
spark.sql("""INSERT INTO employee VALUES(101,"Sweekar",19,500000),(102,"harshal",20,40000)""")
```

```python
spark.sql("""SELECT * FROM employee""").show()
```
