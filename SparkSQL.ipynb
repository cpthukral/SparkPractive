```python
import pyspark
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("SparkSQL").getOrCreate()
```

```python
spark
```

```python
data = [
    (1, "Tanya", 21, "IT", 30000),
    (2, "Vernika",20, "CS", 40000),
    (3, "Aryan", 20, "CS", 50000)
]
```

```python
columns = ["id", "name", "age", "department", "salary"]
```

```python
spk_df = spark.createDataFrame(data,columns)
```

```python
spk_df
```

```python
spk_df.show()
```

```python
spk_df.createOrReplaceTempView("employees")
```

```python
myquery_selectAll = spark.sql("""SELECT * FROM employees""")
myquery_selectAll.show()
```

```python
myquery_Where = spark.sql("SELECT * FROM employees WHERE age > 20")
myquery_Where.show()
```

```python
myquery_OrderBy = spark.sql("SELECT * FROM employees ORDER BY name DESC")
```

```python
myquery_OrderBy.show()
```

```python
spark.sql("SELECT department FROM employees GROUP BY department").show()
```

```python
spk_df.printSchema()
```

```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

schema = StructType([
    StructField("dept_id", IntegerType(), True),
    StructField("dept_name", StringType(), True),
    StructField("course_name", StringType(), True)
    ])
```

```python
my_data = [
    (1, "IT", "B.TECH"),
    (2, "Business","MBA")
]
```

```python
spk_df_schema = spark.createDataFrame(my_data, schema)
```

```python
spk_df_schema.show()
```

```python
spk_df_schema.printSchema()
```

```python
spark = SparkSession.builder.appName("HiveExample").enableHiveSupport().getOrCreate()
```

```python
spark.sql("CREATE TABLE employees (id INT, name STRING, age INT, salary DOUBLE )")
```

```python
spark.sql("DROP VIEW IF EXISTS employees");
spark.sql("INSERT INTO employees VALUES (1, 'John', 25, 50000.0), (2, 'Mary', 30, 60000.0)")
```

```python
df = spark.sql("""
    SELECT *
    FROM employees
    WHERE salary > 55000
""")

df.show()
```
